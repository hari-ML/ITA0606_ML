## Aim
Implement and demonstrate the **FIND-S** algorithm for finding the most specific hypothesis consistent with a set of training examples.

## Algorithm
1. Initialise `h` to the most specific hypothesis `<0, 0, ..., 0>`.
2. For each **positive** training example `x`:
   - for every attribute `a_i`: if `h[i] == 0` set `h[i] = x[i]`; else if `h[i] != x[i]` set `h[i] = ?`.
3. **Negative** examples are ignored.
4. Output `h`.

```python
import pandas as pd

# EnjoySport training data -> stored in a .CSV file
rows = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same',   'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same',   'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes'],
]
cols = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport']
pd.DataFrame(rows, columns=cols).to_csv('enjoysport.csv', index=False)

data = pd.read_csv('enjoysport.csv')
data
```

```python
def find_s(df):
    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values
    n = X.shape[1]
    h = ['0'] * n                       # most specific hypothesis
    print('Initial hypothesis h0 =', h, '\n')
    for i, (x, label) in enumerate(zip(X, y), start=1):
        if label == 'Yes':
            for j in range(n):
                if h[j] == '0':
                    h[j] = x[j]
                elif h[j] != x[j]:
                    h[j] = '?'
            print(f'Example {i}: {list(x)} -> Yes (positive) -> h{i} = {h}')
        else:
            print(f'Example {i}: {list(x)} -> No  (negative, ignored) -> h{i} = {h}')
    return h

final_h = find_s(data)
print('\nMost specific hypothesis found by FIND-S:')
print(final_h)
```

```python
# Use the learned hypothesis to classify new instances
def classify(h, x):
    return 'Yes' if all(hv == '?' or hv == xv for hv, xv in zip(h, x)) else 'No'

tests = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Cool', 'Change'],
    ['Rainy', 'Cold', 'Normal', 'Light',  'Warm', 'Same'],
]
for t in tests:
    print(t, '->', classify(final_h, t))
```

## Result
FIND-S ignores the negative example and generalises only as much as the positive examples require. The final hypothesis `<Sunny, Warm, ?, Strong, ?, ?>` is the **most specific** hypothesis consistent with the data.
